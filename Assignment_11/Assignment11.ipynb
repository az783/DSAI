```python
# Week 11: NLP Assignment
# Software E-Commerce Project
# NLP Preprocessing using TF-IDF (No NLTK)

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# -----------------------------
# 1. Sample Software E-Commerce Reviews Dataset
# -----------------------------
data = {
    "review": [
        "This software is very easy to use and helpful",
        "Excellent features and great performance",
        "Poor customer support and many bugs",
        "Very bad experience, software crashes often",
        "Affordable price and good quality software"
    ]
}

df = pd.DataFrame(data)

print("Original Reviews:\n")
print(df)

# -----------------------------
# 2. TF-IDF Vectorization
#    (includes tokenization + stopword removal)
# -----------------------------
vectorizer = TfidfVectorizer(
    stop_words="english",   # removes stopwords
    lowercase=True          # converts text to lowercase
)

tfidf_matrix = vectorizer.fit_transform(df["review"])

# -----------------------------
# 3. Convert TF-IDF to DataFrame
# -----------------------------
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print("\nTF-IDF Feature Matrix:\n")
print(tfidf_df)

# -----------------------------
# 4. Conclusion
# -----------------------------
print("\nNLP preprocessing completed successfully!")
```
