```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("cleaned_ecommerce.csv")

print("Dataset Shape:", df.shape)
print("\nColumns in Dataset:\n", df.columns.tolist())
```

```python
num_df = df.select_dtypes(include=['int64', 'float64'])

print("\n📈 Basic Statistics Summary:")
for col in num_df.columns:
    print(f"\n➡️ {col}")
    print(f"  Mean: {num_df[col].mean():.2f}")
    print(f"  Median: {num_df[col].median():.2f}")
    print(f"  Mode: {num_df[col].mode()[0]:.2f}")
    print(f"  Variance: {num_df[col].var():.2f}")
```

```python
target_var = 'Rating' if 'Rating' in df.columns else num_df.columns[-1]

# Compute correlation matrix
corr_matrix = num_df.corr(numeric_only=True)

print("\n\n🔍 Correlation Matrix:")
print(corr_matrix[target_var].sort_values(ascending=False))

# Find top 3 features most correlated with target
top_features = corr_matrix[target_var].sort_values(ascending=False)[1:4]

print("\n🏆 Top 3 Features Most Related to Target Variable:")
print(top_features)
```

```python
print("\n🧠 Report Summary:")
print(f"The target variable chosen is '{target_var}'.")
print("The three most related features to the target variable are:")
for feature, value in top_features.items():
    relation = "positively" if value > 0 else "negatively"
    print(f" - {feature} ({relation} correlated, correlation = {value:.2f})")

print("\n✅ Correlation analysis completed successfully!")
```
